# historical vs astronomical 

- 기원전 0 년은 존재하지 않는다. 
- 기원전 1년이  천문학년도에서는 0 년이기 때문이다. 
- 기원전 10년은 천문학년도에서는 -9 년이 된다. 
    

## historical_to_astronomical(year:int)

- 역사 년도를 천문학년도로 바꿔 준다.
- 년도만 바꾼다. 월, 일은 대상이 아니다.

```python
from historical_dates.calendar_utils import (
    astronomical_to_historical,
    historical_to_astronomical,
)
```

```python
print(historical_to_astronomical(-469))
# -468

print(astronomical_to_historical(-468))
# -469

print(historical_to_astronomical(-1))
# 0

print(astronomical_to_historical(0))

# -1
```

```python
person = {
    "name": "소크라테스",
    "birth_year": -469,
    "death_year": -399,
}

from historical_dates.calendar_utils import (
    historical_to_astronomical,
)

birth = historical_to_astronomical(person["birth_year"])

death = historical_to_astronomical(person["death_year"])

print(birth)
# -468

print(death)
# -398
```
