```python
samples = [
    # -----------------------------------------
    # 1. 기원전 / 서기 표기
    # -----------------------------------------
    "기원전 5000년경",
    "기원전 469년",
    "기원전 1년",
    "기원후 100년",
    "서기 100년",
    "BC 100년",
    "BCE 100년",
    "100 BCE",
    "100 BC",
    "AD 100",
    "100 CE",
    # -----------------------------------------
    # 2. 일반 연도
    # -----------------------------------------
    "1",
    "100",
    "800",
    "1564",
    "2026",
    "-1",
    "-469",
    "-800",
    # -----------------------------------------
    # 3. 연월일 형식
    # -----------------------------------------
    "1564-02-15",
    "1564-2-15",
    "1564/02/15",
    "1564.02.15",
    "1564년 2월 15일",
    "기원전 469년 6월 1일",
    "-469-06-01",
    # -----------------------------------------
    # 4. 불완전한 날짜
    # -----------------------------------------
    "1564-02",
    "1564-00-00",
    "1564-02-00",
    "-469-05-00",
    "2026-12",
    # -----------------------------------------
    # 5. 불확실한 날짜
    # -----------------------------------------
    "1564?",
    "1564~",
    "1564%",
    "기원전 469년경",
    "약 1564년",
    "c. 800 BCE",
    "BC c. 469",
    "1564년경",
    # -----------------------------------------
    # 6. 윤년 및 경계 날짜
    # -----------------------------------------
    "2000-02-29",
    "1900-02-28",
    "2024-02-29",
    "2023-12-31",
    "2000-01-01",
    "기원전 1년 2월 29일",
    # -----------------------------------------
    # 7. 확장 연도
    # -----------------------------------------
    "10000",
    "20001",
    "-12000",
    "20001-01-03",
    # -----------------------------------------
    # 8. 잘못된 날짜
    # -----------------------------------------
    "0000",
    "1564-13-01",
    "1564-02-30",
    "1900-02-29",
    "기원전 0년",
    "잘못된 날짜",
    "",
    None,
]
```

```python
for each in samples:
    date = parse_historical_date(each)

    if date is None:
        print(f"✗ {each!s:<25} -> 파싱 실패")
        continue

    numeric = historical_date_to_number(date)

    try:
        edtf = normalize_historical_date(date)
    except ValueError:
        edtf = "지원하지 않는 EDTF 형식"

    print(f"✓ {each!s:<25} -> {edtf!s:<20} -> {numeric:.6f}")
```
